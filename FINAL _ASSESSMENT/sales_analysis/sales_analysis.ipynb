```python
import pandas as pd
import glob
```

```python
# Step 1: Load and merge multiple CSVs into one DataFrame
files = glob.glob(r"C:\Users\Ashwinbarath\Desktop\data\*.csv")  
df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
```

```python
# Step 2: Clean missing values and remove duplicates
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
# Step 3: Create total_sale_value column
df['total_sale_value'] = df['quantity'] * df['price']
```

```python
# Step 4: Group by store_id and get top 5 stores by revenue
top_stores = df.groupby('store_id')['total_sale_value'].sum().reset_index()
top_stores = top_stores.sort_values(by='total_sale_value', ascending=False).head(5)
```

```python
# Step 5: Export final DataFrame to Excel
output_file = r"C:\Users\Ashwinbarath\Desktop\top_5_stores.xlsx"
top_stores.to_excel(output_file, index=False)

print(f"Top 5 stores by revenue exported to {output_file}")
```
