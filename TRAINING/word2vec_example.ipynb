```python
from gensim.models import Word2Vec
```

```python
#  Prepare a tiny corpus (tokenized)
sentences = [
    ["the", "cat", "sits", "on", "the", "mat"],
    ["the", "dog", "sits", "on", "the", "sofa"],
    ["dogs", "and", "cats", "are", "good", "pets"],
    ["i", "love", "my", "pet", "cat"]
]
```

```python
# Train a Word2Vec model (Skip-Gram)
model = Word2Vec(sentences, vector_size=50, window=2, min_count=1, sg=1)
```

```python
#  Get vector for a word
print("Vector for 'cat':")
print(model.wv['cat'])
```

```python
# Find most similar words
print("\nMost similar to 'cat':")
print(model.wv.most_similar('cat', topn=3))
```
