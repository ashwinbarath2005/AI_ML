```python
import cv2
import numpy as np

# Start video capture
cap = cv2.VideoCapture(0)

# Define HSV color range for WHITE
lower_white = np.array([0, 0, 200])    
upper_white = np.array([180, 40, 255]) 

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Convert BGR frame to HSV
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # Create a mask for white color
    mask = cv2.inRange(hsv, lower_white, upper_white)

    # Apply the mask on original frame
    result = cv2.bitwise_and(frame, frame, mask=mask)

    # Show the original, mask, and result
    cv2.imshow("Original", frame)
    cv2.imshow("Mask (White Regions)", mask)
    cv2.imshow("Detected White Objects", result)

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
